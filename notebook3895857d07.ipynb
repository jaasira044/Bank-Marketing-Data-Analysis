```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import urllib.request
import zipfile

# Download dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00222/bank.zip"
urllib.request.urlretrieve(url, "bank.zip")

# Extract dataset
with zipfile.ZipFile("bank.zip", "r") as z:
    z.extractall("bank_data")

# Load dataset
df = pd.read_csv("bank_data/bank-full.csv", sep=";")

# Subscription visualization
plt.figure(figsize=(7,5))
sns.countplot(data=df, x='y')

plt.title("Customer Term Deposit Subscription")
plt.xlabel("Subscribed (No / Yes)")
plt.ylabel("Number of Customers")
plt.show()

print(df['y'].value_counts())
```

```python
plt.figure(figsize=(10,7))

numeric_df = df.select_dtypes(include='number')
sns.heatmap(numeric_df.corr(),
            annot=True,
            cmap='coolwarm',
            fmt='.2f')

plt.title("Correlation Heatmap of Bank Marketing Dataset")
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.countplot(data=df, y='job', hue='y',
              order=df['job'].value_counts().index)

plt.title("Term Deposit Subscription by Job Category")
plt.xlabel("Number of Customers")
plt.ylabel("Job Category")
plt.legend(title="Subscribed")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(data=df, x='month', hue='y',
              order=['jan','feb','mar','apr','may','jun',
                     'jul','aug','sep','oct','nov','dec'])

plt.title("Monthly Term Deposit Subscription Analysis")
plt.xlabel("Month")
plt.ylabel("Number of Customers")
plt.legend(title="Subscribed")

plt.show()
```

```python
plt.figure(figsize=(9,6))

sns.boxplot(data=df, x='y', y='duration')

plt.title("Call Duration vs Term Deposit Subscription")
plt.xlabel("Subscribed (No / Yes)")
plt.ylabel("Call Duration (Seconds)")

plt.show()
```

```python
plt.figure(figsize=(9,6))

sns.boxplot(data=df, x='y', y='age')

plt.title("Customer Age vs Term Deposit Subscription")
plt.xlabel("Subscribed (No / Yes)")
plt.ylabel("Customer Age")

plt.show()
```

```python
print("BANK MARKETING - KEY FINDINGS")

print("\n1. Total Customers:", len(df))

print("\n2. Subscription Percentage:")
print((df['y'].value_counts(normalize=True) * 100).round(2))

print("\n3. Average Customer Age:")
print(round(df['age'].mean(), 2))

print("\n4. Average Bank Balance:")
print(round(df['balance'].mean(), 2))

print("\n5. Average Call Duration:")
print(round(df['duration'].mean(), 2))

print("\n6. Subscription by Job:")
print(df.groupby('job')['y'].value_counts(normalize=True).unstack().round(2))

print("\nEDA Analysis Completed!")
```
