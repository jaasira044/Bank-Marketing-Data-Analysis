```python
import pandas as pd
import zipfile
import urllib.request

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00222/bank.zip"

urllib.request.urlretrieve(url, "bank.zip")

with zipfile.ZipFile("bank.zip", "r") as zip_ref:
    zip_ref.extractall("bank_data")

df = pd.read_csv("bank_data/bank-full.csv", sep=";")

print("Dataset downloaded successfully!")
print("Dataset Shape:", df.shape)
display(df.head())
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())

# Dataset information
print("\nDataset Information:")
df.info()
```

```python
# Statistical summary
display(df.describe())

# Check outliers using IQR
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    outliers = df[
        (df[col] < Q1 - 1.5 * IQR) |
        (df[col] > Q3 + 1.5 * IQR)
    ]

    print(col, "Outliers:", len(outliers))
```

```python
# Cap extreme outliers using IQR
numeric_cols = ['age', 'balance', 'day',
                'duration', 'campaign', 'previous']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower, upper)

print("Outlier handling completed!")
display(df.describe())
```

```python
# Restore original previous column
# Re-download original dataset
original = pd.read_csv("bank_data/bank-full.csv", sep=";")

df["previous"] = original["previous"]

# Final verification
print("Dataset Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())

print("Data Cleaning Completed!")
```

```python
# Convert categorical columns into numerical values
df_encoded = pd.get_dummies(df, drop_first=True, dtype=int)

print("Preprocessing Completed!")
print("Original Shape:", df.shape)
print("Processed Shape:", df_encoded.shape)

display(df_encoded.head())
```

```python
df_encoded.to_csv("preprocessed_bank_marketing.csv", index=False)

print("Preprocessed dataset saved successfully!")
print("Final Shape:", df_encoded.shape)
```
